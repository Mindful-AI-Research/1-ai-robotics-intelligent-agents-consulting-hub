```python
from gtts import gTTS
from IPython.display import Audio, display
import os
import time

# --- Funções do Assistente ---

def falar(texto):
    """
    Converte o texto em fala (áudio) e o reproduz.
    """
    try:
        # 1. Cria o objeto gTTS
        tts = gTTS(text=texto, lang='pt')

        # 2. Salva o áudio em um arquivo temporário
        audio_file = "resposta.mp3"
        tts.save(audio_file)

        # 3. Exibe e reproduz o áudio no Colab
        display(Audio(audio_file, autoplay=True))

        # Opcional: Pausa para a fala terminar (aproximadamente)
        # Calcula a duração do áudio (método não trivial no colab)
        # Vamos usar uma pausa fixa pequena para simplicidade.
        time.sleep(1.5)

    except Exception as e:
        print(f"Erro ao gerar ou reproduzir áudio: {e}")
    finally:
        # 4. Remove o arquivo temporário
        if os.path.exists(audio_file):
            os.remove(audio_file)

def processar_comando(comando):
    """
    Processa o comando (string) e gera a resposta.
    """
    comando = comando.lower().strip() # Padroniza o comando

    # Respostas e Ações
    if "olá" in comando or "oi" in comando:
        resposta = "Olá! Como posso te ajudar hoje?"

    elif "que horas são" in comando or "me diga a hora" in comando:
        # Importa a biblioteca de tempo aqui, apenas se for usar
        import datetime
        hora_atual = datetime.datetime.now().strftime("%H horas e %M minutos")
        resposta = f"A hora atual é {hora_atual}."

    elif "seu nome" in comando:
        resposta = "Eu sou um Assistente Virtual de demonstração, construído com Python!"

    elif "obrigado" in comando or "valeu" in comando:
        resposta = "De nada! Estou aqui para ajudar."

    elif "sair" in comando or "tchau" in comando or "encerrar" in comando:
        resposta = "Tchau! Tenha um ótimo dia!"
        falar(resposta)
        return "SAIR" # Sinaliza o loop principal para encerrar

    else:
        resposta = "Desculpe, não entendi. Você pode tentar perguntar sobre a hora ou me dizer olá."

    # Fala a resposta gerada
    falar(resposta)

# --- Loop Principal ---

print("--- Assistente Virtual (Demo) ---")
print("Comandos de exemplo: olá, que horas são, seu nome, sair")

# Sauda o usuário no início
falar("Bem-vindo ao Assistente Virtual de Demonstração. Estou pronto para o seu comando.")


while True:
    try:
        # SIMULAÇÃO DE RECONHECIMENTO DE VOZ (Entrada de Texto)
        comando_usuario = input("\nSeu comando (digite): ")

        # Processa e verifica se o comando é de saída
        resultado = processar_comando(comando_usuario)

        if resultado == "SAIR":
            break

    except Exception as e:
        print(f"\nOcorreu um erro no loop principal: {e}")
        break

print("\nAssistente encerrado.")
```
